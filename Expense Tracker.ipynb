```python
class ExpenseTracker:

    def __init__(self):
        self.expenses = []

    # add_expense function
    def add_expense(self):
        print("\n--- Add Expense ---")

        category = input("Enter category: ")
        amount = float(input("Enter amount: "))
        description = input("Enter description: ")

        expense = {
            "category": category,
            "amount": amount,
            "description": description
        }

        self.expenses.append(expense)

        print("Expense added successfully!")


    # view_expense function
    def view_expenses(self):
        print("\n--- All Expenses ---")

        if len(self.expenses) == 0:
            print("No expenses found.")
        else:
            for i, expense in enumerate(self.expenses, start=1):
                print("\nExpense", i)
                print("Category:", expense["category"])
                print("Amount: ₹", expense["amount"])
                print("Description:", expense["description"])


    # Function to calculate total expense
    def total_expense(self):
        total = 0

        for expense in self.expenses:
            total = total + expense["amount"]

        print("\nTotal Expense: ₹", total)


    # category-wise expense
    def category_expense(self):
        print("\n--- Category-wise Expense ---")

        if len(self.expenses) == 0:
            print("No expenses found.")
            return

        categories = {}

        for expense in self.expenses:
            category = expense["category"]
            amount = expense["amount"]

            if category in categories:
                categories[category] = categories[category] + amount
            else:
                categories[category] = amount

        for category, amount in categories.items():
            print(category, ": ₹", amount)


    # search_category expense
    def search_category(self):
        search = input("\nEnter category to search: ")

        found = False

        for expense in self.expenses:
            if expense["category"].lower() == search.lower():
                print("\nCategory:", expense["category"])
                print("Amount: ₹", expense["amount"])
                print("Description:", expense["description"])

                found = True

        if found == False:
            print("No expense found for this category.")


    # delete_function expense
    def delete_expense(self):
        self.view_expenses()

        if len(self.expenses) == 0:
            return

        number = int(input("\nEnter expense number to delete: "))

        if number >= 1 and number <= len(self.expenses):
            self.expenses.pop(number - 1)
            print("Expense deleted successfully!")
        else:
            print("Invalid expense number.")



tracker = ExpenseTracker()


while True:

    print("\n==============================")
    print("       EXPENSE TRACKER")
    print("==============================")

    print("1. Add Expense")
    print("2. View Expenses")
    print("3. Total Expense")
    print("4. Category-wise Expense")
    print("5. Search by Category")
    print("6. Delete Expense")
    print("7. Exit")

    choice = input("\nEnter your choice: ")

    if choice == "1":
        tracker.add_expense()

    elif choice == "2":
        tracker.view_expenses()

    elif choice == "3":
        tracker.total_expense()

    elif choice == "4":
        tracker.category_expense()

    elif choice == "5":
        tracker.search_category()

    elif choice == "6":
        tracker.delete_expense()

    elif choice == "7":
        print("\nThank you for using Expense Tracker!")
        break

    else:
        print("\nInvalid choice. Please try again.")
```
